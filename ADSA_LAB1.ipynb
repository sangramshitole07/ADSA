```python
class Node:
  def __init__(self,data):
    self.data=data
    self.next=None
  #adding data in the nodes
node1=Node(10)
node2=Node(20)
node3=Node(30)
node4=Node(40)

#adding the address to next node
node1.next=node2
node2.next=node3
node3.next=node4

#assigning the node1 as head and current node as node1 so that we can traverse through whole list
head=node1
current=node1

#traverse until current points to none/null , there are two ways ethier use the boolean way or is not None way
while current:  #while current is not None:
  print(current.data , end="->")
  current=current.next
print("NULL")
```

```python
class Node:
  def __init__(self,data):
    self.data=data
    self.next=None

head=Node(10)
head.next=Node(20)
head.next.next=Node(30)
head.next.next.next=Node(40)

current =head
while current:
  print(current.data, end="->")
  current=current.next
print("NULL")
```
