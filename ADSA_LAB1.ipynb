```python
'''Problem Statement

Implement a Food Serving System Using Linked Lists

There are N people waiting to be served at two food counters. The people must be served according to their arrival order, so that no person is given an unfair advantage.

The following conditions apply:

The number of people N will always be even.
There are exactly two food counters.
Both counters have the same serving speed.
The people must be divided equally between the two counters.
The first person should be assigned to Counter 1, the second person to Counter 2, the third person to Counter 1, the fourth person to Counter 2, and so on.
The order of people assigned to each counter must be maintained.
Both counters serve their respective people sequentially and simultaneously.
The entire problem must be implemented using Linked Lists only. Do not use Python's built-in lists, queues, stacks, sets, dictionaries, or other data structures.'''



class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None
        self.tail = None

    # Add a node at the end
    def insert(self, data):
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            self.tail = new_node
        else:
            self.tail.next = new_node
            self.tail = new_node

    # Display the linked list
    def display(self):
        current = self.head

        while current is not None:
            print(current.data, end=" → ")
            current = current.next

        print("None")


# -------------------------------
# 1. Create original linked list
# -------------------------------

n = int(input("Enter number of people: "))

if n % 2 != 0:
    print("Number of people must be even.")
else:

    original = LinkedList()

    for i in range(1, n + 1):
        original.insert("P" + str(i))

    print("\nOriginal Queue:")
    original.display()


    # -------------------------------
    # 2. Create two linked lists
    # -------------------------------

    counter1 = LinkedList()
    counter2 = LinkedList()

    current = original.head
    position = 1

    while current is not None:

        if position % 2 == 1:
            counter1.insert(current.data)
        else:
            counter2.insert(current.data)

        current = current.next
        position += 1


    # -------------------------------
    # 3. Display two counter queues
    # -------------------------------

    print("\nCounter 1 Queue:")
    counter1.display()

    print("\nCounter 2 Queue:")
    counter2.display()


    # -------------------------------
    # 4. Serve people
    # -------------------------------

    print("\nServing Order:")

    current1 = counter1.head
    current2 = counter2.head

    while current1 is not None and current2 is not None:

        print("Counter 1 serves:", current1.data)
        print("Counter 2 serves:", current2.data)

        current1 = current1.next
        current2 = current2.next
```
